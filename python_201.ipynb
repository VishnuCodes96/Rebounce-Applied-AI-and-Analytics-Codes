## Text Article Analyzer

A Python program that reads a text file and generates a basic **text analysis report**.

The program analyzes the contents of a text file and provides information such as:

* Total number of words
* Total number of sentences
* Average word length
* Estimated reading time
* Most frequently used words

This project was created as a Python assignment to practice **functions, loops, lists, dictionaries, list comprehensions, regular expressions, file handling, and exception handling**.

The program uses Python's built-in `re` module for regular expressions, so no external packages need to be installed.

```python
import re
```

### Following function counts how many words are present in the text.
* using regular expressions library, `[^\w\s]` matches anything that is NOT a word character or whitespace.
* `.sub` method will replace any and all punctuation marks with ''

```python
def word_count(text):
    words_list = text.split()
    words = [re.sub(r'[^\w\s]', '', word.lower()) for word in words_list]
    total_words = len(words)
    return total_words
```

### Following function counts the number of sentences in the text.

It considers these characters as sentence endings: `.` `!` `?`

```python
def sentence_count(text):
    sentences_list = re.split(r'[.!?]',text)
    sentences = [sen for sen in sentences_list if sen.strip()]
    sentence_count = len(sentences)
    return sentence_count
```

* `r'[.!?]'` is a pattern reconisable in regular expression library.
* `.split` method will take in the text, and split into list items when it encouters a . or ! or ?
* using list comprehension to make a clean list of sentences.
* `sen.strip()` will take of a empty strings, if present in the string

## Following function calculates the **average number of characters per word**.

```python
def avg_word_length(text):
    words_list = text.split()
    words = [re.sub(r'[^\w\s]', '', word.lower()) for word in words_list]

    total_letters = 0
    for word in words:
        total_letters += len(word)

    average_word_length = total_letters / len(words)
    return round(average_word_length,2)
```

## Following function estimates the minutes it would take an average person to read the text.
* The program assumes a reading speed of:  `200 words per minute`

```python
def avg_reading_time(text):
    reading_time = word_count(text) / 200
    return round(reading_time,2)
```

## Following function determines which words appear most frequently in the text.
* However, it ignores certain common words called **stopwords**. For example, words such as: `the, and, is, a, an, of, etc.,` excluded.
* This makes the results more useful because extremely common grammatical words don't dominate the report.

```python
def frequency_calculator(text):
    stopwords = {
        "the", "and", "is", "a", "an", "of", "in", "to", "for", "on", "at", "by","has", "had", "as"
    }
    words_list = text.split()
    words = [re.sub(r'[^\w\s]', '', word.lower()) for word in words_list]

    filtered_words = []
    for word in words:
        if word not in stopwords:
            filtered_words.append(word)

    frequency = {}
    for word in filtered_words:
        if word in frequency:
            frequency[word] += 1
        else:
            frequency[word] = 1

    sorted_words = sorted(frequency.items(), key=lambda item: item[1], reverse=True)
    return sorted_words
```

* stopwords are stored in a set, because it is useful for checking whether a word exists in a collection, using `not in` conditional statement in a `for` loop.
* The function creates a empty dictionary called 'frequency', to store the 'word' and its 'count', after looping through everyword, and increasing its count by 1, if the word has repeated itself anywhere in the text.

## The following function is the one that acts a manager and coordinates the entire Program

The other functions each perform one task, while `main()`:

1. Opens the file.
2. Reads the text.
3. Calls each analysis function.
4. Stores the results.
5. Prints the final report.
6. Handles the situation where the file doesn't exist.

```python
def main(file_path):

    try:
        with open(file_path) as file:
            text = file.read()

        word_count_var = word_count(text)
        sentence_count_var = sentence_count(text)
        avg_word_length_var = avg_word_length(text)
        avg_reading_time_var = avg_reading_time(text)
        frequency_calculator_var = frequency_calculator(text)
        print("TEXT ANALYSIS REPORT")
        print("=" * 55)
        print(f"Total Words           : {word_count_var}")
        print(f"Total Sentences       : {sentence_count_var}")
        print(f"Average Word Length   : {avg_word_length_var} characters")
        print(f"Estimated Read Time   : {avg_reading_time_var} minutes")
        print("\nTop 10 Most Common Words")
        print("-" * 35)
        for word, count in frequency_calculator_var[:10]:
            print(f"{word}\t: {count}")
        print("=" * 55)
    except FileNotFoundError:
        print("Please Check The File Path")
```

* The frequency function returns `(word, count)` tuple. To display the 10 most common words used in the article, I used **List Slicing** and unpacked the tuple through `for` loop.

```python
main("block_of_text.txt")
```
